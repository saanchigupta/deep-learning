```python
import pandas as pd
df = pd.read_csv("/content/multiple_linear_regression_dataset.csv")
df
```

```python
print(df.head())
print(df.columns)
print(df.shape)
```

```python
X = df[["age","experience"]]
y= df["income"]
```

```python
import numpy as np
n_features = X.shape[1]
print(n_features)
w= np.zeros(n_features)
b= 0.0
```

```python
def predict(X,w,b):
    """
    X: input features
    w: weights
    b: bias
    """
    y_pred= X.dot(w)+b
    return y_pred
```

```python
def mean_squared_error(y,y_pred):
    """
    y: actual values
    y_pred: predicted values
    """

    loss= ((y_pred-y)**2).mean()
    return loss
```

```python
def compute_gradients(X,y,y_pred):
    N= len(y)

    #Gradient wrt weights
    dw= (2/N)*X.T.dot(y_pred-y)

    #Gradient wrt bias
    db= (2/N)*(y_pred-y).sum()
    return dw,db
```

```python
def compute_gradients(X,y,y_pred):
    N= len(y)

    #Gradient wrt weights
    dw= (2/N)*X.T.dot(y_pred-y)

    #Gradient wrt bias
    db= (2/N)*(y_pred-y).sum()
    return dw,db
```

```python
def update_parameters(w,b,dw,db,lr):
    w=w-dw*lr
    b=b-db*lr
    return w,b
lr=0.0001
epochs=100000
```

```python
for epoch in range(epochs):
    y_pred= predict(X,w,b)
    loss= mean_squared_error(y,y_pred)
    dw,db= compute_gradients(X,y,y_pred)
    w,b = update_parameters(w,b,dw,db,lr)
    print(y_pred)
    print(w,b)
    print("Loss: ",loss)
```
